## Matrix wrappers

```python
from functools import partial

from ledmap import legacy as matrix
```

```python
base = matrix.Matrix(3, 4)
wrappers = [
    None,
    matrix.Wrapper,
    matrix.FlipLR,
    matrix.FlipUD,
    matrix.Serpentine,
    matrix.Transpose,
    matrix.Rot90,
    matrix.Rot180,
    matrix.Rot270,
    partial(matrix.Limit, last=10),
]
```

```python
for w in wrappers:
    m = base if w is None else w(base)
    print(f"# {m.__class__.__name__}: {m.__class__.__doc__.split('\n', 1)[0].strip()}")
    m.print(sep="  ")
    print()
```
