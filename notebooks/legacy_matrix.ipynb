## 2D LED Matrix

```python
from ledmap.legacy import Custom, Limit, Matrix, Rot90, Serpentine, Transpose
from ledmap.util import get_string
```

## Simple matrix

```python
m = Matrix(3, 4)
```

```python
m.print()
```

```python
m.to_array()
```

```python
m.ledmap()
```

```python
# for WLED
print(get_string(m.dump))
```

```python
# Convert to custom (round-trip)
c = Custom(**m.ledmap())
m.print()
```

## Transposed matrix

```python
m = Transpose(Matrix(3, 4))
```

```python
m.print()
```

```python
m.to_array()
```

```python
m.ledmap()
```

## Complex matrix

```python
m = Limit(Rot90(Serpentine(Matrix(3, 4))), first=1, last=10)
```

```python
m.print()
```

```python
m.to_array()
```

```python
m.ledmap()
```

## Arbitrary matrix

```python
m = Custom([14, 13, 12, 10, 9, 8, 7, 6, -1, 5, 4, 3, -1, -1, 2, 1, -1, -1, -1, 0], 4)
```

```python
m.print()
```

```python
m.to_array()
```

```python
m.ledmap()
```
