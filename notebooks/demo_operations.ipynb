# Array operations

Examples of operations applied to pixel arrays.

```python
import numpy as np

from ledmap.pixels import Mapping, limit, offset, select
```

## 2D Matrix

```python
from ledmap.matrix import reorient, serpentine
```

```python
m = Mapping.from_shape((3, 4))
m
```

```python
# TopLeft
m.apply(reorient, "top-left")
```

```python
# TopRight
m.apply(reorient, "top-right")
```

```python
# BottomLeft
m.apply(reorient, "bottom-left")
```

```python
# BottomRight
m.apply(reorient, "bottom-right")
```

```python
# Transpose
m.apply(np.transpose)
```

```python
# Serpentine
m.apply(serpentine)
```

```python
# FlipLR
m.apply(np.fliplr)
```

```python
# FlipUD
m.apply(np.flipud)
```

```python
# Rot90
m.apply(np.rot90)
```

```python
# Rot180
m.apply(np.rot90, k=2)
```

```python
# Rot270
m.apply(np.rot90, k=3)
```

```python
# Offset
m.apply(offset, -2)
```

```python
# Limit
m.apply(limit, 1, 10)
```

```python
# Select
m.apply(select, m.array % 2 == 0)
```
