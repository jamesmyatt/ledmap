## 2D matrix of pixels

```python
import numpy as np

from ledmap.matrix import make_matrix, serpentine
from ledmap.pixels import Mapping
from ledmap.util import get_string
```

## Simple matrix

```python
m = Mapping.from_shape((3, 4))
m
```

```python
m.array
```

```python
m.to_wled()
```

```python
# for WLED
print(get_string(m.dump_wled))
```

## Transposed matrix

```python
m = Mapping.from_shape((3, 4), vertical=True)
m
```

```python
# Alternative
m.equals(make_matrix(4, 3).T)
```

```python
m.to_wled()
```

## Serpentine matrix

```python
Mapping.from_shape((3, 4)).apply(serpentine)
```

```python
Mapping.from_shape((3, 4), vertical=True).apply(serpentine, vertical=True)
```

## Complex matrix

```python
Mapping.from_shape((3, 4)).apply(serpentine).apply(np.rot90).apply(
    lambda x: np.where((x >= 1) & (x <= 10), x, -1)
)
```

```python
Mapping.from_shape((10, 4), vertical=True).apply(np.roll, 1, axis=0).apply(
    lambda x: x[:5, :]
).apply(np.fliplr)
```

## Arbitrary matrix

```python
m = Mapping.from_list(
    [14, 13, 12, 10, 9, 8, 7, 6, -1, 5, 4, 3, -1, -1, 2, 1, -1, -1, -1, 0],
    shape=(-1, 4),
)
m
```

```python
m.to_wled()
```
