## 2D LED Matrix

```python
import numpy as np

from ledmap import make_array
from ledmap.matrix import make_matrix, serpentine
from ledmap.util import get_string
from ledmap.wled import dump as dump_wled
from ledmap.wled import to_dict as to_wled
```

## Simple matrix

```python
m = make_matrix(3, 4)
```

```python
m
```

```python
to_wled(m)
```

```python
# for WLED
print(get_string(lambda f: dump_wled(m, f)))
```

## Transposed matrix

```python
m = make_matrix(3, 4, vertical=True)
```

```python
# Alternative
np.array_equal(m, make_matrix(4, 3).T)
```

```python
m
```

```python
to_wled(m)
```

## Serpentine matrix

```python
make_matrix(3, 4, serpentine=True)
```

```python
make_matrix(3, 4, serpentine=True, vertical=True)
```

## Complex matrix

```python
m = np.rot90(serpentine(make_matrix(3, 4)))
m = np.where((m >= 1) & (m <= 10), m, -1)
```

```python
m
```

## Arbitrary matrix

```python
m = make_array(
    [14, 13, 12, 10, 9, 8, 7, 6, -1, 5, 4, 3, -1, -1, 2, 1, -1, -1, -1, 0], width=4
)
```

```python
m
```

```python
to_wled(m)
```
